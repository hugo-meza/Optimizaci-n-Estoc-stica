```python
import random
import numpy as np
import matplotlib.pyplot as plt
```

```python
P = [[0.8, 0.2],[0.6, 0.4]]
```

```python
n = 1000
m = 1000
c = [0]
mean = []


for j in range(m):
  for i in range(n):
    a = random.random()
    if a > P[c[-1]][0]:
      c.append(1)
    else:
      c.append(0)
  frec = np.mean(c)
  mean.append(frec)
print("-"*25)
print(f"Probabilidad de lluvia: {round(np.mean(mean), 4)}")

plt.plot(mean)
plt.show()
```

```python
P = [[0.8, 0.2],[0.6, 0.4]]
n = 1000
m = 100000
c = [0]
mean = []


for j in range(m):
  for i in range(n):
    a = random.random()
    if a > P[c[-1]][0]:
      c.append(1)
    else:
      c.append(0)
      break
  mean.append(i+1)
print("-"*25)
print(f"Tiempo medio de retorno a seco: {round(np.mean(mean), 4)}")

c.append(1)
mean2 = []

for j in range(m):
  for i in range(n):
    a = random.random()
    if a > P[c[-1]][0]:
      c.append(1)
      break
    else:
      c.append(0)
  mean2.append(i+1)
print("-"*25)
print(f"Tiempo medio de retorno a lluvioso: {round(np.mean(mean2), 4)}")

```
