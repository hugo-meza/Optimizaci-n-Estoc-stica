```python
# Simulación

for i in range(len(R_matriz)):
  t_f[i] = 1.1 + 0.9 * R_matriz[i]
  if i == 0:
    t_ll[i] = 0
    hora_llegada[i] = 0
    hora_salida[i] = t_f[i]
  else:
    t_ll[i] = -2 * math.log(1 - R_matriz[i])
    hora_llegada[i] = hora_llegada[i - 1] + t_ll[i]
  hora_salida[i] = max(hora_llegada[i], hora_salida[i-1])  + t_f[i]
```

```python
# Mostrar resultados

df = pd.DataFrame(
    {
        "Cliente": clientes,
        "R": R_matriz,
        "t_ll": np.round(t_ll, 4),
        "t_f": np.round(t_f, 4),
        "Hora de LLegada": np.round(hora_llegada, 4),
        "Hora de Salida": np.round(hora_salida, 4),
    }
)

print(df.head(60).to_string(index=False))
```
