```python
import matplotlib.ticker as mticker

def evento(n):
  op = [-1, 1]
  prob = [0.4, 0.6]
  resultado =[]
  resultado.append(0)

  for i in range(n):
      resultado.append(resultado[i] + random.choices(op, prob)[0])

  esperanza = n * 0.2
  print(f"Esperanza teórica: {esperanza}")
  print(f"Esperanza experimental: {np.mean(resultado)}")

  print(f"Varianza teórica: {4 * n * 0.6 * 0.4}")
  print(f"Varianza experimental: {np.var(resultado)}")

  plt.figure(figsize=(10, 6)) # Set figure size for better readability
  plt.plot(resultado, marker='o', linestyle='-', color='b', linewidth=2, label='Paseo Aleatorio') # Plot with markers, line style, color, and thickness
  plt.axhline(y=esperanza, color='r', linestyle='--', label=f'Esperanza ({esperanza:.2f})') # Add horizontal line for esperanza
  plt.title(f"Paseo Aleatorio para n = {n}") # Add a dynamic title
  plt.xlabel("Pasos") # Add x-axis label
  plt.ylabel("Posición") # Add y-axis label
  plt.grid(True) # Add grid
  plt.gca().xaxis.set_major_locator(mticker.MultipleLocator(1)) # Ensure integer x-axis ticks
  plt.gca().yaxis.set_major_locator(mticker.MultipleLocator(1)) # Ensure integer y-axis ticks
  plt.legend() # Show legend
  plt.show() # Display the plot
```

```python
m = [50, 50, 50, 50, 50]
for i in m:
  evento(i)
```

## Probabilidad de regreso a la posición de origen

```python
def regreso():
  n = 100
  op = [-1, 1]
  prob = [0.4, 0.6]
  resultado =[]
  resultado.append(0)
  count = 0
  for i in range(n):
      resultado.append(resultado[i] + random.choices(op, prob)[0])
      count += 1
      if resultado[i+1] == 0:
        # print("Éxito")
        return count
        break
  # print("Fracaso")
  return 0
```

```python
p = 0.6
q = 1 - p
m = 50
exitos = 0
pasos = []
for i in range(m):
  k = regreso()
  if k != 0:
    exitos += 1
    pasos.append(k)

print("Frecuencia relativa experimental: ", exitos/m)
print("Frecuencia relativa teórica: ", 1 - abs(p - q))
print("Media de pasos: ", np.mean(pasos))
print("Media de pasos teórica: ", 1/abs(p - q))
```
