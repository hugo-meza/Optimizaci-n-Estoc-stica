```python
def criterio(x_i, y_i):
  if x_i <= 5 and (y_i >= 2 and y_i <= 3):
    return True
  elif (x_i >= 1 and x_i <= 7) and (y_i >= 1 and y_i <=2):
    return True
  elif (y_i >= 0 and y_i <= 1) and (x_i >= 2 and x_i <= 3)or (x_i >= 5 and x_i <=6):
    return True
  elif (y_i >= 3 and y_i <= 4) and (x_i >= 2 and x_i <= 4) :
    return True
  else:
    return False
```

```python
n = 1000
X = 0
for i in range(len(R_matriz) - 1):
  x_i = R_matriz[i] * 7
  y_i = R_matriz[i + 1] * 4
  if criterio(x_i, y_i):
    X += 1

print("Área aproximada: ", X/len(R_matriz) * 28)
```

# Problema 2

```python
def u():
  return random.random()
```

```python
n = 1000
X = 0
for _ in range(n):
  u_i = u()
  X += (1/(1 + u_i) + 1/(2 - u_i))/2

print("Área aproximada: ", X / n)
print("Área real: ", math.log(2))
print("Error absoluto: ", abs(math.log(2) - X/n))
```
